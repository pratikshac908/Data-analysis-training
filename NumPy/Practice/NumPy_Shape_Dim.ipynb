## What is numpy?

NumPy is a Python library mainly used for scientific and numerical computing. It provides arrays and many useful functions for performing mathematical and logical operations quickly. It is also useful for data analysis, statistics, linear algebra, and random number operations.

The main object in NumPy is called ndarray. It is used to store multidimensional arrays, usually containing elements of the same data type.

## Numpy Arrays Vs ordered groups of values
Fixed Size: NumPy arrays have a fixed size when they are created. If we change the size, a new array is created.

Same Data Type: All elements in a NumPy array usually have the same data type and size.

Fast Operations: NumPy makes mathematical operations on large amounts of data faster and easier with less code.

Widely Used: Many scientific and data-related Python libraries use NumPy arrays for processing data.

### np.ceil() → Ceiling

Definition: Rounds each number UP to the nearest integer.

Always goes toward +∞, no matter what.

```python
import numpy as np

arr = np.array([1.2, 2.7, -1.5, -2.3])
arr
```

```python
np.ceil(arr)
```

### np.floor() → Floor

Definition: Rounds each number DOWN to the nearest integer.

Always goes toward -∞.

```python
np.floor(arr)
```

### np.round() → Round (to nearest)

Definition: Rounds each number to the nearest integer.

If decimal = 0.5 → rounds to nearest even integer (NumPy behavior).

```python
np.round(arr)
```
